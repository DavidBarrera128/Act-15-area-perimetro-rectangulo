## Act 15 área perímetro rectángulo

```python
# 1. Función para calcular el área del rectángulo
def area_rectangulo(base, altura):
    area = base * altura
    return area

# 2. Función para calcular el perímetro del rectángulo
def perimetro_rectangulo(base, altura):
    perimetro = 2 * (base + altura)
    return perimetro

# 3. Función principal del Menú
def menu():
    # --- REDACCIÓN DEL PROBLEMA ---
    print("==================================================")
    print("               PROBLEMA A RESOLVER")
    print("==================================================")
    print("Crear un programa que calcule el área o el perímetro")
    print("de un rectángulo a partir de su base y altura dadas.")

    # --- FÓRMULAS A UTILIZAR ---
    print("\n--- Fórmulas a utilizar ---")
    print("-> Área = base * altura")
    print("-> Perímetro = 2 * base + 2 * altura (o 2*(base+altura))")

    # --- IMAGEN DEL RECTÁNGULO (Arte ASCII) ---
    print("\n--- Imagen del rectángulo ---")
    print("       _________________________")
    print("      |                         |")
    print("      |                         |  Altura")
    print("      |_________________________|")
    print("                 Base")
    print("==================================================\n")

    # Pedir los datos al usuario usando input()
    # Usamos float() para que acepte números con punto decimal
    base = float(input("Por favor, introduce la medida de la BASE: "))
    altura = float(input("Por favor, introduce la medida de la ALTURA: "))

    # Iniciamos un ciclo para mantener el menú activo hasta que elija salir
    while True:
        print("\n--- MENÚ ---")
        print("1. Calcular área del rectángulo")
        print("2. Calcular perímetro del rectángulo")
        print("3. Salir")
        print("4. Mostrar información del creador y salir")

        # Guardamos la opción que elija el usuario
        opcion = input("\nElige una opción (escribe 1, 2, 3 o 4): ")

        # Evaluamos la opción elegida
        if opcion == '1':
            resultado_area = area_rectangulo(base, altura)
            print(f"\n✅ RESULTADO: El área del rectángulo es {resultado_area}")

        elif opcion == '2':
            resultado_perimetro = perimetro_rectangulo(base, altura)
            print(f"\n✅ RESULTADO: El perímetro del rectángulo es {resultado_perimetro}")

        elif opcion == '3':
            print("\n" + "-"*30)
            print("Programa terminado")
            print("Creador del software por David Barrera NL = 8")

        elif opcion == '4':
            print("\n" + "-"*30)
            print("Programa terminado")
            print("Creador del software por David Barrera NL = 8")
            print("\n¡Gracias por usar el programa! Saliendo...")
            break # El break sirve para romper el ciclo y terminar el programa
            break

        else:
            # Por si el usuario teclea algo distinto a 1, 2, 3 o 4
            print("\n❌ Opción no válida. Por favor intenta de nuevo.")

# Para que el programa funcione, necesitamos "llamar" a la función menú al final:
menu()
```
